## Compare Time Series between pre-Omicron and post-Omicron

```python
# Dependencies 
import pandas as pd 
import re
```

```python
# Assign the directory path for the figures
FIGS_PATH='/your_path'
# Read your csv files
pre_tweets = pd.read_csv('Raw_tweets_plus_sentiments_pre.csv', low_memory=False);
post_tweets = pd.read_csv('Raw_tweets_plus_sentiments_post.csv', low_memory=False);
```

```python
pre_tweets.sample()
```

```python
# Create a function to get year and month data from the "created_at" column 
def get_year_month( row ):
    return( row.created_at[:7])
```

```python
# Apply the function to your dataframe to crate a YearMonth column
pre_tweets['YearMonth']  = pre_tweets.apply( get_year_month, axis=1);
post_tweets['YearMonth']  = post_tweets.apply( get_year_month, axis=1);
```

```python
pre_tweets.sample(10)
```

```python
post_tweets.sample(10)
```

### Visualize Time series for pre and post-Omicron

```python
# Plot the time series for the pre-Omicron phase
import plotly.express as px
import pandas as pd

fig = px.line(pre_tweets.groupby('YearMonth').agg(
                Positive=('Sentiment', lambda x: (x == 'positive').sum()),
                Negative=('Sentiment', lambda x: (x == 'negative').sum()),
                Neutral=('Sentiment', lambda x: (x == 'neutral').sum())
).reset_index(), x='YearMonth', y=['Positive','Negative', 'Neutral'], title='pre-Omicron Time Series')

fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1m", step="month", stepmode="backward"),
            dict(count=6, label="6m", step="month", stepmode="backward"),
            dict(count=1, label="YTD", step="year", stepmode="todate"),
            dict(count=1, label="1y", step="year", stepmode="backward"),
            dict(step="all")
        ])
    )
)
fig.write_html("pre-Omicron-TimeSeries.html")
fig.show()              
```

```python
# Plot the time series for the post-Omicron phase

fig = px.line(post_tweets.groupby('YearMonth').agg(
                Positive=('Sentiment', lambda x: (x == 'positive').sum()),
                Negative=('Sentiment', lambda x: (x == 'negative').sum()),
                Neutral=('Sentiment', lambda x: (x == 'neutral').sum())
).reset_index(), x='YearMonth', y=['Positive','Negative', 'Neutral'], title='post-Omicron Time Series')

fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1m", step="month", stepmode="backward"),
            dict(count=6, label="6m", step="month", stepmode="backward"),
            dict(count=1, label="YTD", step="year", stepmode="todate"),
            dict(count=1, label="1y", step="year", stepmode="backward"),
            dict(step="all")
        ])
    )
)
fig.write_html("post-Omicron-TimeSeries.html")
fig.show()              
```
